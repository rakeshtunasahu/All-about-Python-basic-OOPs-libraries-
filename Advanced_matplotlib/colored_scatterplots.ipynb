```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
iris=pd.read_csv(r"C:\Users\ASUS\OneDrive\Desktop\iris.csv")
iris
```

```python
iris.head()
```

```python
plt.scatter(iris['SepalWidthCm'],iris['PetalWidthCm'])
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
```

```python
plt.scatter(
    iris['SepalLengthCm'],
    iris['PetalLengthCm'],
    c=iris['Species'].astype('category').cat.codes,
    cmap='winter',
    alpha=0.98
)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
```
