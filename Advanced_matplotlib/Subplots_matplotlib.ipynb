```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
batters=pd.read_csv(r"C:\Users\ASUS\OneDrive\Desktop\batter (1).csv")
batters
batters.head()
```

```python
plt.figure(figsize=(15,6))
plt.scatter(batters['avg'],batters['strike_rate'])
plt.title('Something')
plt.xlabel('Avg')
plt.ylabel('Strike Rate')

plt.show()
```

```python
fig,ax = plt.subplots(figsize=(15,6))

ax.scatter(batters['avg'],batters['strike_rate'],color='red',marker='+')
ax.set_title('Something')
ax.set_xlabel('Avg')
ax.set_ylabel('Strike Rate')

fig.show()
```

# batter dataset

```python
fig, ax = plt.subplots(nrows=2,ncols=1,sharex=True,figsize=(10,9))

ax[0].scatter(batters['avg'],batters['strike_rate'],color='red',marker='s')
ax[1].scatter(batters['avg'],batters['runs'],marker='+')

ax[0].set_title('Avg Vs Strike Rate')
ax[0].set_ylabel('Strike Rate')


ax[1].set_title('Avg Vs Runs')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
```

```python
fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(10,10))

ax[0,0].plot(batters['avg'],batters['runs'])
ax[0,1].scatter(batters['avg'],batters['runs'])
ax[1,0].hist(batters['avg'])
ax[1,1].hist(batters['runs'])
```

```python
fig = plt.figure()

ax1 = fig.add_subplot(2,2,1)
ax1.hist(batters['avg'], bins=20, color='skyblue', edgecolor='black')

ax2 = fig.add_subplot(2,2,2)
ax2.pie(batters['runs'])

ax3 = fig.add_subplot(2,2,3)
ax3.plot(batters['avg'])
```
